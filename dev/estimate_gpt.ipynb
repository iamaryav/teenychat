```python
import torch
seq_len = 1024
head_dim = 128
base = 10000

t = torch.arange(seq_len)
c = torch.arange(0, head_dim, 2)
inv_freq = 1.0 / (base ** (c / head_dim))
x = torch.outer(t, inv_freq)
print(t.shape, c.shape, x.shape)
print([2], inv_freq[2], x[1:2,:])
x[None, :, None, :].shape
```
